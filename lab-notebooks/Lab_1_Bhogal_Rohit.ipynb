## Basic Data Structures

There are multiple data structures in Python. Some of the common ones include:

    - Lists  = []
    - Tuples = ()
    - Dictionaries = {}
    

### Lists

`Lists = []` are a collection of items that are ordered, mutable, and may have duplicates.

List are indexed and the index always begins at zero `[0, 1, 2, 3]`

```python
lst1 = [1, 2, 3] # Initialize a list with three elements inside
print(lst1)
```

```python
lst1.append('Spartan')
print(lst1)
```

### Tuples

`Tuples = ()` are a collection of items that are ordered and immuatble (cannot be edited). Tuples are processed faster than lists, so they should be used when items do not need to be changed.

```python
t1 = (1,2,3,4,5)
a=6
b=7
c=8
d=9
e=10
t2 = (a,b,c,d,e)

print(t1)
print(t2)
```

```python
#Tuples can be indexed like lists
print(t1[0])
print(t2[2])
```

```python
t1.append(2) # Tuples are immutable - can't change it =(
```

### Dictionaries

`Dictionaries = {}` are ordered, mutable sets of (key:value) pairs. Keys must be unique within a dictionary.

```python
dct1 = {} # initialize an empty dictionary

dct1[1] = "The value is 1" # add a key-value by using bracket (key)
dct1['cat'] = 'meow' # data types of key:values pairs do not have to match
print(dct1)
```

```python
dct1['cat'] # indexing based on a key
```

```python
subject_dct = {}

subject_dct['Subject1'] = ['Male', 'Control', 200] # here the value type is a list
subject_dct['Subject2'] = ['Female', 'Experimental', 100]

subject_dct
```

Dictionaries are mutable

```python
D1 = {'John':117, 'Alice':249, 'George':762} # dictionaries can be initialized empty or like this
print(D1['Alice'])

D1['Quin']=462 #add value
D1['George']=654 #update value
D1.pop('Alice') #deletes 'Alice' entry from D1

print(D1)
```
