Conditionals are used to check whether a condition is True or False and allow the program to react accordingly. The simplest form is the if statement.

```python
cat_count = 4

if cat_count == 0:
    print('Mice are safe!')
```

The if, else statement works when you want one thing to happen when a condition it true, and something else to happen when it is false.

```python
cat_count = 4

if cat_count == 0:
    print('Mice are safe!')
else:
    print('😸😸 Meow Meow')
```

Sometimes there are more than two possibilities. One way to express a decision tree like that is a chained conditional, using elif. Remember `if` always comes first, `elif` in the middle, and `else` is last.

```python
cat_count = -2 # Try this with different numerical values

if cat_count == 0:
    print('Mice are safe!')
elif cat_count > 0:
    print('😸😸 Meow Meow')
else:
    print('How do you have a negative amount of cats?')
```

One conditional can also be nested within another.  However, nested conditionals  very quickly become difficult to read. In general, it is a good idea to avoid too many of them when you can.

```python
x=15#initial value

if x<20:  #if statement
    print('Value is:', x)

print('\n')    
correct_value = True
if type(x) is int:
    print('Acceptable value is:', x)
else:
    print("Value of x is unacceptable")
    correct_value = False

print('\n')        # newline character prints new lines
if correct_value:
    if x>20:  #chained conditional
        print("Value %s is greater than 20"%x)
    elif x<10:
        print("Values %s is less than 10"%x)
    else:
        print('The value', x, 'is between 10 and 20')
```

### try, except

Try, except statements are useful for when something goes wrong.

```python
cat_count = 'dog' # Something is wrong here

try:

    if cat_count == 0:
        print('Mice are safe!')
    elif cat_count > 0:
        print('😸😸 Meow Meow')
    else:
        print('How do you have a negative amount of cats?')

except Exception as e:

    print(e)
```

## Iterations

Computers can automate repetitive tasks. Repeating identical or similar tasks without making errors is something that computers do well.

There are two (main) types of loops.  The first uses a `for` statement to iterate over a defined amount of loops.  The second, a `while` loop, iterates until a condition is met.  While loops can iterate indefinately if not used correctly. -Be careful!!!

### For Loops

```python
# Use the range function to print out a number of specific n

loop_count=0

for i in range(5):
    print(i) #reports the current number of iterations
    loop_count +=1 # increase the value of the loop counter by 1

print(f'The number of loops is: {loop_count}')
```

```python
range(5)
```

```python
for number in range(3, 6):
    print(number)
```

```python
for index in range(2, 10): #Notice where in index starts and stops
    print(index)
```

```python
# Populating a dictionary through a for loop
dct1 = {}

for index in range(1, 10):
    dct1[index] = f'The value is {index}'

print(dct1)
```

```python
#for loop demo
#general case:  for loop_variable in sequence:
x=1
y=2
z=3
 
list_one= [x,y,z]

loop_count=1

for list_item in list_one:  #this for loop is going to iterate through the entire list, so 3 times  
    print("Loop # %s \nItem %s"%(loop_count, list_item)) 
    loop_count=loop_count+1                                   

print('\n')  

#using for with range() 

for i in range(10):    #this is loops through a range of 0-9 (10 items), and is a very useful form of for loop utilization
    print(i)
```

### While Loops

```python
count = 0
while (count < 9):
    print('The count is:', count)
    count += 1
```

```python
conditional = True
x = 0
while conditional:
    print(f"x = {x}")
    if x>=30:
        conditional = False
    x+=3
```

```python
conditional = True
x = 0
while conditional:
    print(f"x = {x}")
    if x>=30:
        conditional = False
    x+=3
```

```python
# this loop would run forever, but uses a break statement when a condition is met to end the loop
y = 0
while True:
    print(f"y = {y}")
    if y>=20:
        break # ends the loop
    y+=2

# Be careful to avoid infinite loops as they can crash your computer!!!
```
