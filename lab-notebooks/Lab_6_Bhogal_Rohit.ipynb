## What is NumPy?

**NumPy** stands for **Numerical Python**.

It is a Python library that gives us a special data type called an **array**. A NumPy array is similar to a Python list, but it is designed specifically for numbers and numerical operations.

That means NumPy makes it easier to:
- store many values together
- perform math on all values at once
- organize data into rows and columns
- prepare for future work in data science, engineering, biology, healthcare, and machine learning

---

## How does NumPy build from what we already know?

You already know how to make a Python list like this:

```python
heart_rates = [72, 88, 95, 67, 81]
```

A NumPy array looks very similar:

```python
heart_rates = np.array([72, 88, 95, 67, 81])
```

So the idea is familiar:
- both store multiple values
- both use square brackets
- both can be indexed

But NumPy arrays are much stronger for number-based work because they support fast, clean **element-wise math**.

For example, with NumPy you can add 5 to every value in the array with one line of code.

## Step 2: Creating a 1D NumPy Array

A **1D array** is like a single row of values.

This is similar to a Python list, but created with `np.array()`.

### Example


````python
heart_rates = np.array([72, 88, 95, 67, 81])
print(heart_rates)

````

### What happened?

We created a NumPy array named `heart_rates`.

Notice the pattern:
- `np.array(...)` creates the array
- the values go inside square brackets
- the result prints a sequence of numbers

This should feel similar to a Python list, which is good. NumPy is building on something you already know.


## Step 3: Python List vs NumPy Array

Let's compare them directly.


````python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
````

At first glance, these look similar.

But the real difference appears when we do math.


````python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print("Python list * 2 =", python_list * 2)
print("NumPy array * 2 =", numpy_array * 2)
````

### Important idea

A Python list repeated by 2 gives:

```python
[1, 2, 3, 1, 2, 3]
```

A NumPy array multiplied by 2 gives:

```python
[2 4 6]
```

That is because NumPy applies the math to **each value** in the array.

This is called **element-wise operation**.


## Step 4: Inspecting Array Information

NumPy arrays have useful properties that help us understand the data.


````python
heart_rates = np.array([72, 88, 95, 67, 81])

print("Array:", heart_rates)
print("Shape:", heart_rates.shape)
print("Number of dimensions:", heart_rates.ndim)
print("Size:", heart_rates.size)
print("Data type:", heart_rates.dtype)
````

### Vocabulary

- `shape` tells us the structure of the array
- `ndim` tells us how many dimensions it has
- `size` tells us the total number of values
- `dtype` tells us the kind of data stored inside

For a simple 1D array of 5 values, the shape will look like `(5,)`.


## Step 5: Indexing a NumPy Array

Just like Python lists, NumPy arrays use indexing.

Remember:
- index `0` is the first value
- index `1` is the second value
- index `-1` is the last value


````python
heart_rates = np.array([72, 88, 95, 67, 81])

print("First value:", heart_rates[0])
print("Third value:", heart_rates[2])
print("Last value:", heart_rates[-1])
````

## Step 6: Element-Wise Math

This is one of the biggest reasons NumPy is so useful.

You can perform math across the whole array at once.


````python
heart_rates = np.array([72, 88, 95, 67, 81])

print("Add 5:", heart_rates + 5)
print("Multiply by 2:", heart_rates * 2)
print("Subtract 10:", heart_rates - 10)
````

Each operation is applied to every value in the array.

This saves time and makes code cleaner.


## Step 7: Common NumPy Functions

NumPy also includes built-in functions for common numerical summaries.


````python
heart_rates = np.array([72, 88, 95, 67, 81])

print("Sum:", np.sum(heart_rates))
print("Mean:", np.mean(heart_rates))
print("Max:", np.max(heart_rates))
print("Min:", np.min(heart_rates))
````

These functions are especially useful when working with measurements, scores, sensor readings, lab data, or patient data.


## Step 8: Creating a 2D Array

A **2D array** is like a table with rows and columns.

This is useful for storing grouped data.
For example, each row could represent a patient, and each column could represent a measurement.


````python
patient_data = np.array([
    [72, 98.6, 120],
    [88, 99.1, 135],
    [95, 100.2, 140]
])

print(patient_data)
````

### Thinking about the table

You can imagine this as:

- row 0 -> first patient
- row 1 -> second patient
- row 2 -> third patient

and:

- column 0 -> heart rate
- column 1 -> temperature
- column 2 -> systolic pressure


````python
print("Shape:", patient_data.shape)
print("Dimensions:", patient_data.ndim)
````

A shape of `(3, 3)` means:
- 3 rows
- 3 columns


## Step 9: Indexing a 2D Array

For a 2D array, indexing uses:

```python
array_name[row, column]
```

The first number is the row.
The second number is the column.


````python
print(patient_data[0, 0])   # first row, first column
print(patient_data[1, 1])   # second row, second column
print(patient_data[2, 2])   # third row, third column
````

## Step 10: Simple Slicing

Slicing lets us grab part of an array.

For a 1D array:


````python
heart_rates = np.array([72, 88, 95, 67, 81])
print(heart_rates[1:4])
````

For a 2D array:
- `:` means all rows or all columns in that position


````python
print("First row:", patient_data[0, :])
print("First column:", patient_data[:, 0])
````

# Data Analysis Exercises
Background: Diffusion and Fick's Law

In biotransport, molecules often move across membranes through **diffusion**.

One way to estimate the diffusion rate is using **Fick's Law**:

$
J = -D \frac{dC}{dx}
$

where:

- **J** = diffusion flux (mol/m²·s)
- **D** = diffusion coefficient (m²/s)
- **dC/dx** = concentration gradient across the membrane


## Experimental Data

Distance across membrane (meters):

0.0, 0.002, 0.004, 0.006, 0.008

Measured oxygen concentration (mol/m³):

0.90, 0.75, 0.62, 0.51, 0.40

````python
import numpy as np

distance = [0.0, 0.002, 0.004, 0.006, 0.008]
concentration = [0.90, 0.75, 0.62, 0.51, 0.40]

D = 2.1e-9

````

# Exercise 4: Diffusion Flux Calculation (3 pts)

Convert the lists to NumPy arrays and calculate the diffusion flux across the membrane.

### Part A (1 pt)
Convert `distance` and `concentration` into **NumPy arrays**.

### Part B (1 pt)
Calculate the **concentration gradient** using:


$dC/dx = \frac{C_{last} - C_{first}}{x_{last} - x_{first}}$

Print the gradient.

### Part C (1 pt)
Using **Fick's Law**, calculate the diffusion flux:

$J = -D \frac{dC}{dx}$

Print the flux value.

````python
# Part A
d = np.array(distance)
c = np.array(concentration)

# Part B
conc_gradient = (c[-1] - c[0]) / (d[-1] - d[0])

# Part C
flux = -D*conc_gradient
print(flux)

````
