## Classes

Classes store data and functions. They allow you to:

* Model real-world problems, while hiding the details with abstraction
* Re-use code
* Encapsulate related data and behavior

Classes are defined using the *class* keyword. An example definition of a Car class is given below.

```python
class Car:
    def __init__(self, make, model, year, color):
        self.make = make
        self.model = model
        self.year = year
        self.color = color
        self.started = False
        self.speed = 0
        self.max_speed = 200

    def start(self):
        print("Starting the car... vroom, vroom!")
        self.started = True

    def stop(self):
        if self.speed > 40:
            print("Whoa, slow down first or you'll get whiplash!")
        else:
            print("Stopping the car...")
            self.started = False

    def accelerate(self, value):
        if not self.started:
            print("Car is not started!")
            return
        if self.speed + value <= self.max_speed:
            self.speed += value
        else:
            self.speed = self.max_speed
        print(f"Accelerating to {self.speed} mph...")

    def brake(self, value):
        if self.speed - value >= 0:
            self.speed -= value
        else:
            self.speed = 0
        print(f"Braking to {self.speed} mph...")

    def __str__(self):
        return f"This is a {self.color} {self.make} {self.model} ({self.year})"
```

Classes contain data, also called *attributes* and functions, which in the context of classes are called *methods*. The Car class above contains the attributes make, model, year, color, started, speed and max speed. It has the methods start, stop, accelerate, brake. An *object* is an instance of the class with specific values for its attributes. To create an object of a class, you use the class name and a parenthesized list of values for its attributes as allowed by its \_\_init\_\_() method. See the first line in the code below for an example. Once you instantiate an object of a class such as the mustang object of the Car class below, you can access its other methods such as `start()`, `stop()`, `accelerate()`, and `brake()` using dot notation as shown in the next code block. The special \_\_str\_\_ method is a method called when the object needs to be type cast to a string, such as in a print statement.

```python
mustang = Car('Ford', 'Mustang Boss 429', '1969', 'Red') # Instantiation: creating and specific instance of the Car class

# Examples of dot notation
mustang.start()
mustang.accelerate(35)
mustang.brake(10)
mustang.stop()

print(mustang) # Returns the string from the __str__ function
```

We can create multiple objects from the same class. We can create as many objects of our class as we need. To create another car object that is a Toyota Corolla 2025, we can follow a similar structure as before with the new arguments for the *constructor* \_\_init\_\_ method of our car class as follows.

```python
newcar = Car('Toyota', 'Corolla', '2025', 'Black')
print(newcar)
```

## Inheritance
Inhertiance allows us to create a new *child* class extending an existing *parent* class by adding new attributes and methods. For instance, we can extend our Car class to a HybridCar class as follows.  

```python
class HybridCar(Car):
    def __init__(self, make, model, year, color, battery_voltage):
        super().__init__(make, model, year, color)
        self.battery_voltage = battery_voltage
    def get_battery_voltage(self):
        return self.battery_voltage
    def __str__(self):
        return f"This is a {self.color} {self.make} {self.model} ({self.year}) and has a battery voltage of {self.battery_voltage} volts"
```

The super function calls the constructor of the parent class. We can instantiate a new object from the child class like that of the parent class:

```python
my_hybrid_car = HybridCar('Toyota', 'Camry', '2026', 'Black', 252)
print(my_hybrid_car)
```

Notice how both Car and Hybrid Car had methods called \_\_str\_\_. This is a type of *method overriding* where a child class provides another implementation of a parent's method. When the method is called on an object of the child class, the implementation in the child class is used. Otherwise, the method in the parent class is used as the following code illustrates. 

```python
my_hybrid_car.start() 
```

## Modules

In Python, a module or package is a collection of pre-defined functions and classes that can be used to save time.  Python packages add a great deal of functionality.   Standard modules are included with python, and don't have to be installed separately.  

In later lessons you will learn how to manage packages that are installed on your device, but you will find that the Anaconda Distribution that you installed comes with many of the most useful packages included.  

Here we see examples of some useful modules. For more information regarding Modules and Packages, I recommend:

[General Module info](https://docs.python.org/3/tutorial/modules.html)

One useful module is the *time* module. Before using the time module's functions can be used you need to `import` the module.  Reasoning:

(1) It lets Python know which package you are talking about

(2) It lets you specify how much of the package to load into memory

Functions and classes available in a module are accessed using a dot notation. 

```python
# Example adapted from W3schools: https://www.w3schools.com/python/ref_module_time.asp

import time
start = time.time()
print(f'Start time: {start}')
time.sleep(5) # Code wil take at least 5 seconds to run
end = time.time()
print(f'End time: {end}')
print(f'Elapsed: {end - start:.2f} seconds')
```

The time() function in the time module returns the number of seconds since January 1, 1970. To get a more readable time, you can use the ctime function.

```python
time.ctime()
```

Another useful module is the [math](https://docs.python.org/3/library/math.html) module that provides access to common mathematical functions. For instance, to compute the number of bacteria after 5 rounds of binary division, we can take 2 to the power 5 using: 

```python
import math

num_bacteria = math.exp2(5)
print(f'The number of bacteria after 5 rounds is: {num_bacteria}')
```

The math module also contains some of the common constants. To compute the area of a Petri Dish of 50 mm radius, we can do so as follows:

```python
import math
petri_area = math.pi*math.pow(50, 2) # pow(50,2) = 50 to the power of 2
print(f'The area of a Petri Dish of 50 mm radius is: {petri_area}')
```

To save time and space (memory) you can import individuals functions using the `from module_name import function_name` construct as shown in the code below. You can also import every function using the \* wildcard. However, avoid using `from module_name import *` as this loads the whole module to memory, more memory = slower in some situations, and can cause errors.

```python
from math import pi, pow
petri_area = math.pi*math.pow(50, 2) # pow(50,2) = 50 to the power of 2
print(f'The area of a Petri Dish of 50 mm radius is: {petri_area}')
```
