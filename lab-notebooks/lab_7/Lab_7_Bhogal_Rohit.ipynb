# Using Pandas Examples

## Step 1: Import Pandas

Make sure you have pandas installed. If you installed Anaconda, pandas should have been installed automatically. If you are using another python package manager like pip, refer to the instructions at https://pandas.pydata.org/docs/getting_started/index.html for installation instructions.

Before we can use Pandas, we need to import it.

Run the cell below.

````python
import pandas as pd

print("Pandas loaded successfully")
````

## Step 2: What is Pandas?

Pandas is a Python library used to work with **tabular data**.

A Pandas table is called a **DataFrame**.

You can think of a DataFrame like:
- an Excel sheet
- a table with rows and columns
- a CSV file loaded into Python

## Step 3: Creating a Simple DataFrame

We can build a DataFrame from a Python dictionary.
Each key becomes a column.

````python

lab_data = {
    "Sample": ["Control", "Treatment A", "Treatment B"],
    "Abs600": [0.62, 0.81, 0.77],
    "Replicate": [1, 1, 1]
}

df = pd.DataFrame(lab_data)
print(df)
````

### What happened?

We created a DataFrame named `df`.

Notice:
- column names are at the top
- each horizontal line is a row
- each row stores one observation

## Step 4: Viewing the First Rows

A common beginner function is:

```python
df.head()
```

This shows the first 5 rows by default.

````python
df.head()
````

## Step 5: Inspecting Basic Information

When you first load a dataset, you usually want to ask:
- how many rows and columns are there?
- what are the column names?
- what type of object is this?

````python

print("Type:", type(df))
print("Shape:", df.shape)
print("Columns:", df.columns)
````

### Vocabulary

- `DataFrame` = a table of data
- `shape` = number of rows and columns
- `columns` = column names

## Step 6: Selecting One Column

To select one column, use:

```python
df["column_name"]
```

This returns a **Series**, which is like one column of values.

````python
print(df["Abs600"])
````

## Step 7: Selecting More Than One Column

Use a list of column names inside double square brackets.

````python
print(df[["Sample", "Abs600"]])
````

## Step 8: Selecting Rows with `iloc`

`iloc` uses integer positions.

- row 0 = first row
- row 1 = second row
- row -1 = last row

````python

print("First row:")
print(df.iloc[0])

print("\nLast row:")
print(df.iloc[-1])
````

## Step 9: Simple Summary Functions

Pandas can quickly summarize numeric data.

````python

print("Mean Abs600:", df["Abs600"].mean())
print("Max Abs600:", df["Abs600"].max())
print("Min Abs600:", df["Abs600"].min())
````

These functions are useful for lab data because they help us summarize measurements without doing every calculation by hand.

## Step 10: Reading a CSV File

One of the most common Pandas tasks is loading data from a CSV file.

The basic syntax is:

```python
pd.read_csv("file_name.csv")
```

````python

plate_df = pd.read_csv("plate_reader_day1.csv")
plate_df
````

## Step 11: Looking at the CSV Data

After reading a CSV, it is good practice to inspect it right away.

````python

print("Shape:", plate_df.shape)
print("Columns:", list(plate_df.columns))
print()
print(plate_df.head())
````

## Step 12: Selecting Useful Columns

Plate reader files often contain many columns.
You will usually focus on the columns that matter for your analysis.

````python
print(plate_df[["Well", "Sample", "Abs600"]])
````

## Step 13: Filtering Rows

We can keep only rows that match a condition.

For example, here we keep rows where the `Condition` is `Treatment`.

````python

treatment_rows = plate_df[plate_df["Condition"] == "Treatment"]
print(treatment_rows)
````

## Step 14: Combining Pandas with Skills You Already Know

Pandas still connects to core Python ideas you already know.

For example:
- variables
- strings
- printing
- indexing
- conditionals

````python

threshold = 0.75
mean_treatment = treatment_rows["Abs600"].mean()

print("Threshold:", threshold)
print("Treatment mean:", mean_treatment)

if mean_treatment > threshold:
    print("Treatment mean is above the threshold.")
else:
    print("Treatment mean is not above the threshold.")
````

## Exercise 4: Filtering and Comparing Groups (3 pts)

Use the file `plate_reader_day2.csv`.

### Part A (1 pt)
Read the CSV file into a DataFrame called `plate2`.

Create:
- `control_df` containing only rows where `Condition` is `"Control"`
- `treatment_df` containing only rows where `Condition` is `"Treatment"`

### Part B (1 pt)
Calculate and print:
1. the mean `Abs600` for the control group
2. the mean `Abs600` for the treatment group

### Part C (1 pt)
Use an `if/else` statement to compare the two means and print:
- `"Treatment mean is higher"` if the treatment mean is larger
- `"Control mean is higher"` otherwise

````python
# Part A
plate2 = pd.read_csv('plate_reader_day2.csv')
control_df = plate2[plate2["Condition"] == "Control"]
treatment_df = plate2[plate2["Condition"] == "Treatment"]

# Part B
control_mean = control_df["Abs600"].mean()
treatment_mean = treatment_df["Abs600"].mean()
print('Control Mean:', control_mean)
print('Treatment Mean:', treatment_mean)
print()

# Part c
if treatment_mean > control_mean:
    print("Treatment mean is higher")
else:
    print("Control mean is higher")
````
